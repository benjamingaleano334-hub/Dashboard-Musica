**Idea**: Armar una lista de familias, cada una con palabras clave. Si el nombre de un género contiene una palabra clave, pertenece a esa familia. Después, a cada artista se le asigna la familia que más se repite entre sus géneros.

```python
# Reglas
reglas = [
    ("Hip hop / rap", ["hip hop", "rap", "trap", "drill", "grime"]),
    ("R&B / soul / funk", ["r&b", "soul", "funk", "motown", "disco", "quiet storm"]),
    ("Electrónica", ["edm", "house", "techno", "electro", "trance", "dubstep",
                     "dance", "synth", "drum and bass", "electronic", "downtempo", "ambient"]),
    ("Metal / punk", ["metal", "punk", "hardcore", "grunge", "emo"]),
    ("Rock", ["rock", "new wave", "britpop", "shoegaze", "mellow gold", "indie"]),
    ("Country / folk", ["country", "folk", "bluegrass", "americana", "singer-songwriter"]),
    ("Jazz / blues", ["jazz", "blues", "swing", "bebop", "big band", "boogie", "ragtime"]),
    ("Latina", ["latin", "salsa", "reggaeton", "bachata", "tango", "mexican", "cumbia",
                "ranchera", "mariachi", "bolero", "merengue", "samba", "bossa", "flamenco"]),
    ("Reggae / caribeña", ["reggae", "dancehall", "ska", "soca"]),
    ("Clásica", ["classical", "orchestra", "opera", "baroque", "romantic", "symphony",
                 "choral", "piano", "chamber", "soundtrack"]),
    ("Pop", ["pop", "adult standards", "easy listening", "lounge", "vocal", "crooner", "teen"]),
    ("Mundo / regional", ["greek", "arab", "indian", "bollywood", "persian", "turkish",
                          "african", "afro", "k-pop", "j-pop", "anime", "celtic", "gospel"]),
]
```

```python
# Funciones
from collections import Counter

def familia_de_genero(genero):
    for familia, palabras in reglas:
        for palabra in palabras:
            if palabra in genero:
                return familia
    return None
```

Recorre las familias en orden. Si una palabra clave aparece dentro del nombre del género, devuelve esa familia. Si ninguna coincide, devuelve None (nada).

```python
def familia_de_artista(lista_generos):
    if len(lista_generos) == 0:
        return "Sin género"
    familias = [familia_de_genero(g) for g in lista_generos]
    familias = [f for f in familias if f is not None]
    if len(familias) == 0:
        return "Otros"
    return Counter(familias).most_common(1)[0][0]
```

```python
# Aplicar la regla
generos_limpio["familia"] = generos_limpio["genres"].apply(familia_de_artista)

generos_limpio["familia"].value_counts()
```

```python
df_final = df_limpio.merge(
    generos_limpio[["artists", "familia"]],
    left_on="artista_principal",
    right_on="artists",
    how="left",
)

df_final["familia"] = df_final["familia"].fillna("Sin género")
df_final = df_final.drop(columns=["artists"])

print(len(df_limpio), len(df_final))
```

Qué hace, en orden:
- `merge` une las tablas haciendo coincidir `artista_principal` con `artists`. Esta vez traés solo las dos columnas que importan (`artists` y `familia`).
- `how="left"` conserva todas las canciones, aunque su artista no aparezca en `generos`.
- `fillna("Sin género")` completa con ese texto las canciones cuyo artista no se encontró (eran solo unas 12).
- `drop(columns=["artists"])` borra la columna `artists`, que quedó duplicada con `artista_principal` después del cruce.
- El `print` compara las filas antes y después: **los dos números tienen que ser iguales** (165.424). Si `df_final` tuviera más, el cruce habría duplicado canciones.

## Agrupación de géneros en familias

### Por qué agrupar
El archivo `data_w_genres.csv` tiene 2.972 géneros distintos (`classic rock`, `folk rock`, `vocal jazz`, `mellow gold`...). Son demasiados para graficar y muchos son variantes de lo mismo. Por eso los agrupé en familias.

### Cuántas familias definí
Definí **12 familias**:
Hip hop / rap, R&B / soul / funk, Electrónica, Metal / punk, Rock, Country / folk, Jazz / blues, Latina, Reggae / caribeña, Clásica, Pop y Mundo / regional.

Además de esas 12, hay dos etiquetas especiales: **"Otros"** y **"Sin género"**, que explico más abajo.

### Criterio de asignación
1. **Palabras clave.** Cada familia tiene una lista de palabras clave. Si el nombre de un género contiene una de ellas, pertenece a esa familia. Por ejemplo, `classic rock`, `hard rock` y `folk rock` contienen "rock", así que van a Rock.
2. **Orden de prioridad.** Las familias se revisan en el orden en que las definí y **gana la primera que coincide**. Esto resuelve los géneros ambiguos: `pop rap` va a Hip hop / rap (que viene antes que Pop) y `folk rock` va a Rock (que viene antes que Country / folk).
3. **Familia más frecuente por artista.** Un artista puede tener varios géneros. Convierto cada uno en su familia y le asigno al artista la familia que más se repite. Si hay un empate, queda la que apareció primero en su lista.
4. **Cruce con las canciones.** La familia se asigna **por artista** y después se une a las canciones usando el artista principal de cada una. Es decir, todas las canciones de un mismo artista quedan en la misma familia.

### Qué pasa con "Otros" y "Sin género"
Son dos casos distintos:
- **"Otros"**: el artista **sí tiene géneros**, pero ninguno coincide con mis palabras clave. Por ejemplo, géneros como `show tunes`, `broadway`, `children's music` o `chanson`.
- **"Sin género"**: el artista **no tiene ningún género** cargado, o no aparece en el archivo de géneros. Son cerca del 8,4% de las canciones.

Conservo ambas categorías en la tabla para no perder canciones en las secciones de evolución, popularidad y artistas. En los gráficos de géneros las excluyo, para que no se confundan con una familia real.

### Limitaciones
- Las palabras clave y el orden de prioridad son **decisiones mías**, no una clasificación oficial. Otra persona podría armar familias distintas.
- Hay géneros ambiguos. Por ejemplo, `dance pop` cae en Electrónica porque contiene "dance", pero podría ir en Pop.
- Un artista con géneros de varias familias queda en una sola, aunque haga música de más de un estilo.
- Una búsqueda por palabras clave puede clasificar mal algún género raro cuyo nombre contenga una de ellas por casualidad.
